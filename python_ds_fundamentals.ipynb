# Python Data Structures Fundamentals

Reference notes for Python's core built-in collections, before starting LeetCode.
Goal: be able to create, iterate, and reason about time complexity **without looking anything up**.

## 1. Lists (Python's "array")

A `list` is a dynamic array — resizable, ordered, indexable.

### Create

```python
nums = [1, 2, 3]
nums = list()                  # empty list
nums = [0] * 5                 # [0, 0, 0, 0, 0]
nums = list(range(5))          # [0, 1, 2, 3, 4]
squares = [x*x for x in range(5)]   # list comprehension -> [0, 1, 4, 9, 16]
print(nums, squares)
```

### Common operations

```python
nums = [1, 2, 3]
nums.append(4)          # add to end
nums.insert(0, 99)      # insert at index (shifts everything right)
print(nums)

nums.pop()              # remove & return last element
nums.pop(0)             # remove & return element at index 0
print(nums)

nums.remove(3)          # remove first occurrence of VALUE 3
print(nums)

print(nums[0])          # access by index
print(nums[-1])         # last element
print(nums[1:3])        # slicing -> sublist [1, 3)
print(len(nums))
print(3 in nums)        # membership check

nums.sort()             # in-place sort
print(sorted(nums))     # returns new sorted list
nums.reverse()          # in-place reverse
print(nums[::-1])       # reversed copy
```

### Iteration patterns

```python
nums = [10, 20, 30]
list2 = ['a', 'b', 'c']

for x in nums:                      # value only
    print(x)

for i, x in enumerate(nums):        # index + value
    print(i, x)

for a, b in zip(nums, list2):       # parallel iteration
    print(a, b)

for i in range(len(nums)-1, -1, -1):  # reverse index iteration
    print(nums[i])
```

### Time complexity (average case)

| Operation                | Complexity |
|---------------------------|-----------|
| Index access `nums[i]`    | O(1)      |
| Append (end)               | O(1)*     |
| Pop from end                | O(1)      |
| Pop/insert at front or middle | O(n)   |
| Search (`in`, `.index()`)  | O(n)      |
| Sort                        | O(n log n) |

\* O(1) amortized — occasionally a resize costs O(n).

**Key insight:** lists are great for index-based access and stack-like use (end operations), bad for frequent front/middle insertions/removals — use `collections.deque` for that instead.

## 2. Dictionaries (hash maps)

`dict` maps keys to values via hashing. Insertion order is preserved (Python 3.7+).

### Create

```python
d = {}
d = dict()
d = {"a": 1, "b": 2}
d = {k: v for k, v in [("a", 1), ("b", 2)]}   # dict comprehension
print(d)
```

### Common operations

```python
d = {"a": 1, "b": 2}
d["a"] = 1                 # insert or update
print(d["a"])               # access (KeyError if missing)
print(d.get("a"))           # access, returns None if missing
print(d.get("z", 0))        # access with default
print("a" in d)              # membership check (key)

del d["b"]                  # remove key
print(d)

d["a"] = 1
val = d.pop("a")            # remove & return value
print(val, d)

d = {"a": 1, "b": 2}
print(d.pop("z", None))     # remove safely with default
print(list(d.keys()))
print(list(d.values()))
print(list(d.items()))
print(len(d))
```

### Iteration patterns

```python
d = {"a": 1, "b": 2, "c": 3}

for key in d:                    # keys only
    print(key)

for key, value in d.items():     # key + value
    print(key, value)

for value in d.values():         # values only
    print(value)
```

### `collections.Counter` (frequency map — used constantly in LeetCode)

```python
from collections import Counter
c = Counter("aabbbc")        # Counter({'b': 3, 'a': 2, 'c': 1})
print(c)
print(c.most_common(2))      # [('b', 3), ('a', 2)]
print(c["z"])                # 0, not KeyError — defaults missing keys to 0
```

### `collections.defaultdict` (avoids manual key-checking)

```python
from collections import defaultdict
d = defaultdict(list)
d["fruits"].append("apple")   # no need to check key exists first
print(d)
```

### Time complexity (average case)

| Operation       | Complexity |
|-----------------|-----------|
| Insert            | O(1)      |
| Lookup (`in`, `[]`, `.get`) | O(1) |
| Delete             | O(1)      |
| Iteration          | O(n)      |

**Worst case is O(n)** due to hash collisions, but you can treat it as O(1) in practice/interviews unless asked otherwise.

## 3. Sets

A `set` is an unordered collection of unique, hashable elements. Same hash-table backing as `dict`, so same complexity profile.

### Create

```python
s = set()
s = {1, 2, 3}
s = set([1, 2, 2, 3])        # {1, 2, 3} — duplicates removed
s = {x for x in range(5)}    # set comprehension
print(s)
```

### Common operations

```python
s = {1, 2, 3}
s.add(4)
print(s)

s.remove(4)        # KeyError if missing
print(s)

s.discard(99)       # no error if missing
print(4 in s)        # O(1) membership check — this is the #1 reason to use sets
print(len(s))

a = {1, 2, 3}
b = {2, 3, 4}
print(a | b)               # union
print(a & b)               # intersection
print(a - b)               # difference
print(a ^ b)                # symmetric difference
```

### Iteration

```python
s = {1, 2, 3}
for x in s:           # no guaranteed order
    print(x)
```

### Time complexity

| Operation | Complexity |
|-----------|-----------|
| Add        | O(1) |
| Remove      | O(1) |
| Membership (`in`) | O(1) |

**Key insight:** if you ever find yourself writing `if x in some_list` inside a loop, that's an O(n) check repeated n times = O(n²). Converting `some_list` to a `set` first makes it O(n) total. This single trick solves a huge fraction of early LeetCode problems (e.g. Two Sum, duplicate detection).

## 4. Stacks

Python has no dedicated stack type — a `list` works fine since append/pop from the **end** are both O(1).

```python
stack = []
stack.append(1)     # push
stack.append(2)
print(stack.pop())   # pop -> 2 (LIFO: last in, first out)
print(stack[-1])     # peek at top without removing
```

Use case: matching parentheses, DFS, undo functionality, "look back" problems.

## 5. Queues

Don't use a plain list for a queue — `list.pop(0)` is O(n) because everything shifts left. Use `collections.deque` (double-ended queue) instead, which gives O(1) on both ends.

```python
from collections import deque

q = deque()
q.append(1)          # add to right
q.append(2)
print(q.popleft())    # remove from left -> 1 (FIFO: first in, first out)
q.appendleft(0)       # add to left
print(q)
print(q.pop())         # remove from right
```

### `deque` time complexity

| Operation | Complexity |
|-----------|-----------|
| Append/pop (either end) | O(1) |
| Access middle element (`q[i]`) | O(n) |

Use case: BFS, sliding window problems, "process in order" tasks.

## 6. Tuples (bonus — comes up often)

Immutable, ordered. Useful as dict keys (lists can't be dict keys because they're unhashable) and for returning multiple values.

```python
t = (1, 2)
x, y = t                  # unpacking
print(x, y)

d = {(0, 0): "origin"}     # tuple as dict key
print(d)
```

## 7. Quick decision guide

| Need | Use |
|------|-----|
| Indexed, ordered collection | `list` |
| Fast existence/duplicate checks | `set` |
| Key → value lookups | `dict` |
| Frequency counting | `Counter` |
| LIFO (last in, first out) | `list` (as stack) |
| FIFO (first in, first out) | `deque` |
| Fast inserts/removals at both ends | `deque` |
| Immutable / hashable grouped values | `tuple` |
